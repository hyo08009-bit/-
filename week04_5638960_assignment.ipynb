## 문제 1. Tensor 5개 생성과 속성 확인표 (40점)
서로 다른 Tensor 5개를 만들고, 아래 항목을 담은 표를 출력하세요.

| 이름 | shape | dim | dtype | device |

각 Tensor를 왜 그렇게 만들었는지 주석으로 한 줄 적으세요.


```python
# 답안

import torch

# GPU 사용 가능 여부 확인
device = "cuda" if torch.cuda.is_available() else "cpu"

# 1. 정수형 Tensor의 자료형과 구조를 확인하기 위해 생성
a = torch.tensor([1, 2, 3])

# 2. 실수형 Tensor와 정수형 Tensor의 차이를 비교하기 위해 생성
b = torch.tensor([1.0, 2.0, 3.0])

# 3. 모든 원소가 0인 2차원 Tensor를 확인하기 위해 생성
c = torch.zeros(2, 3)

# 4. 표준정규분포 난수로 구성된 3차원 Tensor를 확인하기 위해 생성
d = torch.randn(2, 3, 4)

# 5. GPU 장치에서 생성한 Tensor의 device를 확인하기 위해 생성
e = torch.randn(3, device=device)

# Tensor별 속성 출력
tensors = {"a": a, "b": b, "c": c, "d": d, "e": e}

print("GPU 사용 가능:", torch.cuda.is_available())
print()

for name, t in tensors.items():
    print(f"이름: {name}")
    print(f"shape: {t.shape}")
    print(f"dim: {t.dim()}")
    print(f"dtype: {t.dtype}")
    print(f"device: {t.device}")
    print("-" * 30)
```

## 문제 2. Tensor 연산 (20점)
(3, 4) 와 (4, 2) Tensor를 만들어 아래를 출력하세요.
- 행렬곱 결과와 그 shape
- 첫 번째 Tensor의 열별 평균과 행별 합계


```python
# 답안

import torch

# 1. (3, 4)와 (4, 2) 크기의 Tensor 생성
a = torch.arange(1, 13, dtype=torch.float32).reshape(3, 4)
b = torch.arange(1, 9, dtype=torch.float32).reshape(4, 2)

print("Tensor a:")
print(a)
print("Tensor b:")
print(b)

# 2. 행렬곱 결과와 Shape 확인
result = a @ b

print("\n행렬곱 결과:")
print(result)
print("행렬곱 Shape:", result.shape)

# 3. 첫 번째 Tensor의 열별 평균
col_mean = a.mean(dim=0)
print("\n열별 평균:", col_mean)

# 4. 첫 번째 Tensor의 행별 합계
row_sum = a.sum(dim=1)
print("행별 합계:", row_sum)
```

## 문제 3. CPU와 GPU 비교 (20점)
같은 크기의 행렬곱을 CPU와 GPU에서 각각 실행해 걸린 시간을 출력하고, 차이를 한두 줄로 설명하세요.
GPU를 쓸 수 없는 경우에는 CPU 결과와 그 이유를 적으세요.


```python
# 답안

import torch
import time

# 행렬 크기 설정
n = 2000

# 동일한 데이터를 CPU와 GPU에서 사용
a_cpu = torch.randn(n, n)
b_cpu = torch.randn(n, n)

a_gpu = a_cpu.to("cuda")
b_gpu = b_cpu.to("cuda")

# 사전 연산으로 초기 실행에 따른 영향을 줄임
_ = a_cpu @ b_cpu
_ = a_gpu @ b_gpu
torch.cuda.synchronize()

# CPU 행렬곱 시간 측정
start = time.perf_counter()
result_cpu = a_cpu @ b_cpu
cpu_time = time.perf_counter() - start

# GPU 행렬곱 시간 측정
torch.cuda.synchronize()
start = time.perf_counter()
result_gpu = a_gpu @ b_gpu
torch.cuda.synchronize()
gpu_time = time.perf_counter() - start

# 결과 출력
print(f"CPU 실행시간: {cpu_time:.6f}초")
print(f"GPU 실행시간: {gpu_time:.6f}초")
print(f"CPU / GPU 시간 비율: {cpu_time / gpu_time:.2f}배")
```


# CPU와 GPU 실행시간 비교 결과
# CPU 실행시간: 0.168575초
# GPU 실행시간: 0.006598초
# GPU가 CPU보다 약 25.55배 빠르게 실행되었다.
#
# GPU는 많은 연산을 병렬로 처리할 수 있기 때문에
# 대규모 행렬곱에서 CPU보다 빠른 성능을 보였다.
# 단, 실행시간은 행렬 크기와 하드웨어 환경에 따라 달라질 수 있다.


## 문제 4. 자동미분 (20점)
아래 두 가지를 하세요.
1. $y = 2x^3 + 5x$ 에서 $x=3$ 일 때의 도함수를 자동미분으로 구하고, 손으로 구한 값과 비교하세요.
2. 자기가 만든 수식 하나에 대해 미분값을 구하고, 손 계산과 일치하는지 확인하세요.


```python
# 답안

import torch

# 문제 1. 주어진 수식의 자동미분
x = torch.tensor(3.0, requires_grad=True)
y = 2*x**3 + 5*x

y.backward()

print("문제 1")
print("자동미분 결과:", x.grad.item())
print("손 계산 결과:", 6*3**2 + 5)


# 문제 2. 직접 만든 수식의 자동미분
# y = 3x² + 2x + 1, x = 2
a = torch.tensor(2.0, requires_grad=True)
b = 3*a**2 + 2*a + 1

b.backward()

print("\n문제 2")
print("자동미분 결과:", a.grad.item())
print("손 계산 결과:", 6*2 + 2)
```
